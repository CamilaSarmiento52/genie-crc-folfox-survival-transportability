# Held-out model evaluation and transportability

This notebook generates absolute-risk
predictions for every untouched assessment row from registry-verified trained-model bundles, and
reports delayed-entry-aware discrimination, calibration, and prediction error at 365 and 730 days
with patient-level bootstrap uncertainty. It computes the pooled-versus-held-out transportability gap
and the centre-specific complexity-value table. It generates no explanation, attribution,
prediction-stability, recalibration, or sensitivity result




```python
from pathlib import Path

import pandas as pd
from IPython.display import display

from crc_folfox.evaluate.workflow import run_evaluation_workflow

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
ANALYSIS_DIRECTORY = PROJECT_ROOT / "analysis" / "out" / "current"

result = run_evaluation_workflow(ANALYSIS_DIRECTORY)
```

```python
import json

assert result.artifacts is not None
manifest = json.loads(result.artifacts.evaluation_manifest.read_text(encoding="utf-8"))
keys = ("input_hashes", "software_versions", "row_counts")
display({"workflow": "held-out model evaluation", **{key: manifest[key] for key in keys}})
```

## Evaluation populations, designs, and entry support

Two designs are evaluated for the **same** eligible patients of each centre: `iecv` (the single
institution-held-out model, which never saw that centre) and `pooled` (that patient's pooled
out-of-fold prediction, averaged across admissible repeats). The pooled comparator is a mixed-centre
comparator by design and is never used as a substitute for institution holdout.


```python
display(result.entry_support_summary)
```

## Scope inventory

48 evaluation scopes (2 endpoints x 2 designs x 3 centres x 4 required Cox models) each carry 14
performance rows: two horizon-independent primary rows (truncation-aware concordance, integrated
Brier score), eight horizon-specific primary rows (time-dependent AUC, Brier score, calibration
slope, calibration-in-the-large at 365 and 730 days), and four entry-conditional secondary
calibration rows. `ltrc_rrf` and `ltrc_boosting` were found `technically_ineligible` during model training and
contribute no scope.


```python
inventory = (
    result.performance_summary.groupby(["endpoint", "design", "model"], sort=True)
    .agg(rows=("metric", "count"))
    .reset_index()
)
display(inventory)
```

## Metric set and delayed-entry evidence

Every reported estimator explicitly accepts `entry`, `exit`, and `event` and excludes a patient from
a risk set before entry: the shared delayed-entry censoring KM
(`crc_folfox.evaluate.censoring.fit_censoring_km`/`horizon_weights`), truncation-aware concordance,
Uno-style IPCW time-dependent AUC, the model-training workflow's delayed-entry IPCW Brier core (integrated over
the locked `[365, 730]`-day grid for IBS), and a complementary-log-log-link weighted GLM for the
calibration slope and calibration-in-the-large. `lifelines.utils.concordance_index` and every
`sksurv.metrics.*` survival metric are named ineligible because their interfaces carry no entry time
(`evaluation-spec.md`, "Explicitly ineligible software").

**Decision-curve analysis is prospectively omitted**: no prespecified clinical action or threshold
range is supplied, and the 84-day risk set has only four eligible deaths overall. `net_benefit_gain`
is `not_applicable` in every complexity-value row.


```python
reporting_status_counts = result.performance_summary["reporting_status"].value_counts()
display(reporting_status_counts)
```

## Discrimination, calibration, and prediction-error tables

Every cell carries a point estimate, a 95% bootstrap interval (2000 patient-level resamples per
`(endpoint, centre)`, shared across every design/model/horizon/metric of that cell so comparisons are
exactly paired), support counts, and the prespecified reporting status. A wide interval is
`estimable_imprecise`, never relabelled `not_computable`.


```python
display(
    result.performance_summary[
        [
            "endpoint",
            "design",
            "centre",
            "model",
            "metric",
            "horizon",
            "risk_scale",
            "estimate",
            "ci_lower",
            "ci_upper",
            "n_eligible",
            "n_events",
            "reporting_status",
            "reason",
        ]
    ].head(30)
)
```

## Calibration figures and entry-conditional coherence

Panel figures (one per endpoint/design/horizon, a centre-by-model grid) are saved under
`figures/` from the aggregate quintile calibration curves only -- no patient-level point is plotted.
The entry-conditional secondary risk scale is reported beside the unconditional primary scale as a
prespecified coherence check.


```python
figure_files = sorted(p.name for p in result.artifacts.figures.glob("*.png"))
display(figure_files)

coherence = (
    result.performance_summary[(result.performance_summary["metric"].isin(["citl", "slope"]))]
    .pivot_table(
        index=["endpoint", "design", "centre", "model", "metric", "horizon"],
        columns="risk_scale",
        values="estimate",
    )
    .reset_index()
)
display(coherence.head(20))
```

## Transportability gap

`gap = pooled - iecv` on the same eligible patients and the same paired bootstrap resamples, for the
ten primary metric-horizon combinations. `higher_is_better` is recorded explicitly per metric so
direction is unambiguous rather than left to the reader.


```python
display(
    result.transportability_summary[
        [
            "endpoint",
            "centre",
            "model",
            "metric",
            "horizon",
            "higher_is_better",
            "gap",
            "gap_ci_lower",
            "gap_ci_upper",
            "pooled_status",
            "iecv_status",
        ]
    ]
)
```

## Complexity-value table

For each non-reference candidate (`core_cox`, `elastic_net_cox`, `spline_cox`) compared with
`minimal_cox` on the same eligible patients and paired resamples: seven gains
(`auc_gain` x2, `ibs_gain`, `citl_gain` x2, `slope_gain` x2) and the locked mechanical classification
-- **clear added value**, **trade-off/uncertain**, or **no demonstrated advantage** -- produced
purely from the frozen rule, not a written conclusion. A competitive `minimal_cox` is a substantive
result, not a failed analysis.


```python
classification_by_cell = result.complexity_value_summary.drop_duplicates(
    ["endpoint", "design", "centre", "fuller_model"]
)[["endpoint", "design", "centre", "fuller_model", "classification"]]
display(classification_by_cell)

display(
    result.complexity_value_summary[
        [
            "endpoint",
            "design",
            "centre",
            "fuller_model",
            "gain",
            "horizon",
            "value",
            "ci_lower",
            "ci_upper",
        ]
    ].head(21)
)
```
